# Week 8 Assignment: Single Agent System with Tool Routing

## Objective

The objective of this assignment is to build a simple rule-based single-agent system capable of routing user queries to different tools based on their intent.

The agent supports three types of requests:

- Mathematical calculations
- Keyword extraction
- General conversation

The system analyzes the user query, selects the appropriate tool, and returns the response in a structured JSON format. It also includes validation test cases and an interactive loop for continuous testing.

## Step 1: Import Required Libraries

This step imports the Python libraries required for the assignment.

- **re** is used for regular expression-based text processing.
- **json** is used to format outputs as structured JSON responses.

```python
import re
import json
```

## Step 2: Build the Calculator Tool

The calculator tool evaluates mathematical expressions entered by the user.

If the expression is valid, it returns the calculated result.

If the expression is invalid, an error is generated and handled safely.

```python
def calculator(expression):
    try:
        result = eval(expression)
        return result
    except Exception:
        raise ValueError("Invalid mathematical expression.")
```

## Step 3: Build the Keyword Extraction Tool

This tool extracts meaningful keywords from a given sentence.

The process includes:

- converting text to lowercase
- removing punctuation
- removing common stop words
- returning the remaining important words

```python
def extract_keywords(text):
    words = re.findall(r'\b\w+\b', text.lower())

    stop_words = {
        "the", "is", "and", "a", "an",
        "of", "to", "in", "on", "for",
        "with", "this", "that", "it"
    }

    keywords = [word for word in words if word not in stop_words]

    return keywords
```

## Step 4: Build the Agent Routing Function

The agent acts as the central controller of the system.

It examines the user query and routes it to the appropriate tool.

Routing Rules:

- Queries containing **"calculate"** are sent to the Calculator Tool.
- Queries containing **"keywords"** are sent to the Keyword Extraction Tool.
- All remaining queries receive a general response.

The function also performs error handling and returns every response in JSON format.

```python
def agent(query):

    query_lower = query.lower()

    try:

        # Route to Calculator Tool
        if "calculate" in query_lower:

            expression = query_lower.replace("calculate", "").strip()

            if expression == "":
                return {
                    "type": "error",
                    "result": "No mathematical expression found."
                }

            answer = calculator(expression)

            return {
                "type": "calculation",
                "result": answer
            }

        # Route to Keyword Extraction Tool
        elif "keywords" in query_lower:

            text = query.replace("keywords", "").strip()

            if text == "":
                return {
                    "type": "error",
                    "result": "No text provided for keyword extraction."
                }

            words = extract_keywords(text)

            return {
                "type": "keywords",
                "result": words
            }

        # General Response
        else:

            return {
                "type": "general",
                "result": "I can help with calculations and keyword extraction."
            }

    except Exception as e:

        return {
            "type": "error",
            "result": str(e)
        }
```

## Step 5: Validate the Agent

This step verifies that the routing system works correctly.

Different query types are tested, including:

- calculation requests
- keyword extraction requests
- general conversation
- invalid mathematical expressions
- missing inputs

Each response is displayed in a structured JSON format.

```python
# Validation Test Cases

test_queries = [
    "calculate 5+5",
    "calculate (10+5)*2",
    "keywords Artificial Intelligence is transforming healthcare",
    "Hello",
    "calculate",
    "calculate 5++",
    "keywords"
]

print("===== Validation Results =====\n")

for query in test_queries:
    print(f"Query: {query}")
    response = agent(query)
    print(json.dumps(response, indent=4))
    print("-" * 50)
```

## Step 6: Interactive Testing

This section allows users to interact with the agent by entering queries continuously.

The program accepts user input, routes the query to the appropriate tool, and displays the response in JSON format.

The loop continues until the user enters **exit**.

```python
while True:

    query = input("Enter your query (type 'exit' to quit): ")

    if query.lower() == "exit":
        print("Program terminated.")
        break

    response = agent(query)

    print(json.dumps(response, indent=4))
```
